# OEMOF - a modular open source framework to model energy supply systems
## 1. Intro

**ENG:** Calculating energy systems can quickly become complex. This may be because they involve many time steps, because a network comprising numerous components needs to be modelled, or because energy flows need to be optimised under specific boundary conditions.
There are various frameworks available for this within the Python ecosystem – one of them is oemof, which provides packages for different levels: oemof.network provides the basic framework for mathematical modelling, oemof.solph provides rudimentary components, and oemof.tabular even allows instantiation using specific technical components in *.csv format.

**DE:** Die Berechnung von Energysystemen kann schnell komplex werden. Sei es, weil sie viele Zeitschritte zu berücksichtigen hat, ein Netzwerk aus vielen Komponenten abgebildet werden soll oder weil Energieflüsse unter bestimmten Randbedingungen optimiert werden sollen.
Dafür gibt es im Python Universum verschiedene frameworks - eins davon ist oemof, welches Pakete für verschiedene Level mitbringt: oemof.network stellt das Grundgerüst der mathematischen Modellierung dar, oemof.solph stellt rudimentäre Komponenten bereit und oemof.tabular ermöglicht sogar eine Instanzierung mithilfe von konkreten technischen Komponenten im *.csv -Format.


## 2. main concept of oemof.solph
* readthedocs and be aware of the chosen version: https://oemof-solph.readthedocs.io/en/v0.6.4/
* i suggest to start reading here: https://oemof-solph.readthedocs.io/en/v0.6.4/basic_concepts.html

**EN:** An energy system, defined as an object, contains all the relevant information. This includes details of time and time steps, the components and their connections, as well as any additional constraints, such as limited investment costs or aggregated energy flows.
Solph provides the following components: Sink, Source, Converter, GenericStorage, OffsetConverter, ExtractionTurbineCHP, GenericCHP. The components are connected via ‘trains’, also known as buses. The main purpose of a bus is to balance its inputs and outputs at any given time.

**DE:** Ein Energiesystem als object enthält alle Informationen. Dazu gehören die Angaben zur Zeit und den Zeitschritten, die Komponenten und ihre Verbindungen sowie gegebenenfalls weitere Randbedingungen, wie limitierte Investitionskosten oder aufsummierte Energieströme.
Solph stellt folgende Komponenten bereit: Sink, Source, Converter, GenericStorage, OffsetConverter, ExtractionTurbineCHP, GenericCHP. Die Komponenten sind über "Schienen" aka Bus verbunden. Der Hauptzweck eines Bus ist es, seine Ein- und Ausgänge zu jedem beliebigen Zeitpunkt auszugleichen.

<img src="https://oemof-solph.readthedocs.io/en/v0.6.4/_images/oemof_solph_example.svg">

**EN:** Perhaps the most important component is the ‘Flow’ component. This connects the individual elements (Component <-> Bus or Bus <-> Bus) and is usually defined at the same time as a component is created. In addition, flows contain important variables that the solver optimises. These may include the energy flow per time step or the capacity of a component! Specific boundary conditions (cost per unit of flow, as well as upper or lower limits) are also set there.

**DE:** Die vielleicht wichtigste Komponente ist jedoch die "Flow"-Komponente. Diese verbindet die einzelnen Elemente (Component <-> Bus or Bus <-> Bus) und wird meist beim Erstellen einer Komponente mitdefiniert. Zusätzlich enthalten Flows wichtige Variablen, die der Solver optimiert. Das kann der Energiefluss je Zeitschritt sein oder die Kapazität einer Komponente! Spezifische Randbedingungen (Kosten je einer Einheit des Flows sowie obere oder untere Limits) werden dort ebenfalls gesetzt.
