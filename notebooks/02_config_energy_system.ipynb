## 1. timeseries & timeindex
Every energy system requires a time frame – a sequence of time steps that describe the state of the energy system at any given point in time.
The example shows a time index with 24 hourly time points. Each time point represents one hour of the energy system.

````python
import pandas as pd

# Define simulation period
start_date = "2026-01-01"
periods = 24

# Create time index
timeindex = pd.date_range(
    start=start_date,
    periods=periods,
    freq="h"
)

print(timeindex)
````

## 2. create the energy system
Be aware: it will use your last definition of the timeindex variable!

````python
import oemof.solph as solph

# Create EnergySystem with pandas timeindex
energysystem = solph.EnergySystem(
    timeindex=timeindex,
    infer_last_interval=False
)

# OR Try the internal function
energysystem = solph.EnergySystem(
    timeindex=solph.create_time_index(year=2026,interval=.25,number=24*4)
)

print(energysystem.timeindex)
````
